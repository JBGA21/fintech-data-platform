## Marts

### Stock Daily Performance Data Mart

```python
%sql
CREATE OR REPLACE VIEW fintech.marts.stock_daily_performance AS 

SELECT
    s.symbol,
    d.full_date AS trade_date,

    -- Date attributes
    d.year,
    d.quarter,
    d.month,
    d.month_name,
    d.week_of_year,
    d.day_of_week,
    d.day_name,

    -- Stock prices
    f.open,
    f.high,
    f.low,
    f.close,
    f.volume,
    f.change,
    f.change_percent,
    f.vwap,

    -- Derived performance metrics
    ((f.close - f.open) / f.open) * 100
        AS intraday_return_percent,

    (f.high - f.low)
        AS daily_price_range,

    ((f.high - f.low) / f.low) * 100
        AS daily_range_percent

FROM fintech.gold.fact_stock_prices f

INNER JOIN fintech.gold.dim_stock s
    ON f.stock_key = s.stock_key

INNER JOIN fintech.gold.dim_date d
    ON f.date_key = d.date_key
```
