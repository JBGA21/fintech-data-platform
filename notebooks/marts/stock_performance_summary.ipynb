## Marts

### Stock Performance Summary

```python
%sql
CREATE OR REPLACE VIEW fintech.marts.stock_performance_summary AS 

SELECT 
    symbol,

    -- Date range
    MIN(trade_date) AS first_trade_date,
    MAX(trade_date) AS last_trade_date,

    -- Price metrics
    MIN(close) AS min_close,
    MAX(close) AS max_close,
    AVG(close) AS avg_close,

    -- First / last price
    MIN_BY(close, trade_date) AS first_close,
    MAX_BY(close, trade_date) AS last_close,

    -- Volume metrics
    SUM(volume) AS total_volume,
    AVG(volume) AS avg_daily_volume,

    -- Return metrics
    (
        (
            MAX_BY(close, trade_date)
            - 
            MIN_BY(close, trade_date)
        )
        /
        MIN_BY(close, trade_date)
    ) * 100 AS total_return_percent,

    AVG(change_percent) AS avg_daily_return_percent,

    MAX(change_percent) AS best_day_return_percent,
    MIN(change_percent) AS worst_day_return_percent,

    -- Volatily / range metrics
    AVG(daily_range_percent) AS avg_daily_range_percent,
    MAX(daily_range_percent) AS max_daily_range_percent


FROM fintech.marts.stock_daily_performance

GROUP BY symbol;
```
